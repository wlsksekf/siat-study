```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
# 사용할 번역 모델의 이름
# 이 모델은 "한국어 -> 영어" 번역을 수행하는 인공지능 모델
model_id = "Helsinki-NLP/opus-mt-ko-en"
# 토크나이저를 불러옴
# 토크나이저는 문장을 컴퓨터가 이해할 수 있는 숫자 형태로 바꾸는 도구
tokenizer = AutoTokenizer.from_pretrained(model_id)
# 번역을 수행하는 인공지능 모델을 불러옴
# Seq2Seq 모델은 "문장을 입력받아 새로운 문장을 생성하는" 구조
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)
# 번역할 한국어 문장을 준비
korean_text = "인공지능 공부는 생각보다 정말 재밌어요!"
# 문장을 숫자로 변환(토큰화 과정)
# return_tensors="pt" 는 결과를 PyTorch 형식으로 반환하라는 의미
inputs = tokenizer(korean_text, return_tensors="pt")
# 인공지능 모델이 번역을 수행
# generate() 함수는 입력 문장을 기반으로 새로운 문장을 생성
outputs = model.generate(**inputs)
# 모델이 만든 숫자 결과를 다시 사람이 읽을 수 있는 문장으로 변환
# skip_special_tokens=True 는 특수 기호들을 제거하고 순수 문장만 출력
translated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
# 번역 결과 출력
print(f"한글: {korean_text}")
print(f"영어: {translated_text}")
```
