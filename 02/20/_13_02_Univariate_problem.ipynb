문제1. 취미별 도수분포표 막대그래프와 원그래프로 작성하되 한 화면에 작성하세요

hobby = pd.Series(['등산', '낚시', '골프', '수영', '등산', '등산',
                   '낚시', '수영', '등산', '낚시', '수영', '골프'])

```python
import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams['font.family'] = 'Malgun Gothic' 
plt.rcParams['axes.unicode_minus'] = False 

hobby = pd.Series(['등산', '낚시', '골프', '수영', '등산', '등산',
                '낚시', '수영', '등산', '낚시', '수영', '골프'])

fd = hobby.value_counts()

fig, axes = plt.subplots(1, 2)  

fd.plot.bar(ax=axes[0])
fd.plot.pie(ax=axes[1])

fig.suptitle('선호 취미 분포', fontsize=14)
plt.show()
```

문제2. BostonHousing.csv를 읽어 집값(medv) 데이터 컬럼만 추출하여 house_price에 저장하세요

```python
df = pd.read_csv('data/BostonHousing.csv')

house_price = df['medv']

print(house_price)
```

---
문제3. 집값의 평균, 중앙값, 사분위수를 확인하세요

```python
print(house_price.mean())
print(house_price.median())
print(house_price.quantile([0.25, 0.5, 0.75]))
```

문제4. 사분위수를 기준으로 4개 구간의 집값 평균을 구하고 이를 막대 그래프로 시각화하세요.

- grp = pd.qcut(house_price, q=4, labels=['Q1', 'Q2', 'Q3', 'Q4']) : 데이터를 개수가 거의 동일하도록 자동으로 4개 구간(사분위수) 으로 분할

- house_price.groupby(grp).mean() : 각 사분위(Q1~Q4)에 속한 주택 가격의 평균값 계산

```python
grp = pd.qcut(house_price, q=4, labels=['Q1', 'Q2', 'Q3', 'Q4'])
avg = house_price.groupby(grp).mean()

avg.plot.bar()
plt.xlabel('사분위수')
plt.ylabel('집값 평균')
plt.title('사분위별 집값 평균')
plt.show()
```
