중앙 값의 정의는 데이터 셋의 50% 즉, 말그대로 정 중앙에 있는 값을 의미한다. 그러면 이건 어떻게 구할 수 있는가? 간단한다. 데이터 개수가 n이라면 n에 1을 더한 후에 2를 나누면 된다. 자세한 것은 예제를 통해서 알아보자. 


위의 예제에서 보면 친절하게도 총 데이터가 홀수 인 경우와 짝수 인 경우를 나누어서 설명해주었다.

a는 총 데이터의 개수가 9개인 경우이다. 먼저 데이터 셋을 오름차순으로 쓴 후, (n+1)/2를 계산한다. 즉, (9+1)/2를 하면 5가 된다. 그러면 5번째 값을 오름차순으로 정리한 데이터에서 찾으면 그것이 바로 중앙값(median), M이 되는 것이다. 여기서는 M=3이 되겠다. 

그러면 b를 보면, 

총 데이터의 개수가 10개로 짝수인 경우이다. 이때에는 (n+1)/2를 계산하면 항상 .5가 생길 것이다. 즉, (10+1)/2 = 5.5가 되므로, 5.5번째의 값을 찾아야 하는데, 이 때에는 5번째(4)와 6번째(6) 값을 찾은 후에 그것의 절반을 구해야 한다. 즉, (4+6)/2 = 5로 하여 M = 5가 된다. 

​

자, 그럼 중앙값에 대해서는 이해됐다면 IQR로 넘어가자. 

사분 범위라고 번역이 되는데 이건 대체 머하는 걸까? 일단, 표준편차와 비슷하게 데이터의 변화 정도를 측정한다는 것 정도만 알고 있는데, 정확하게 어떻게 계산하며 그 정의가 어떻게 되는지 확실히 알아보자. 조금 기초적인 부분일지라도 기초가 완벽해야 정말 어려운 것도 이해를 할 수 있으니 기초에 항상 충실하자!


사분 범위 IQR이라는 것의 정의는 

중간에 50%의 데이터들이 흩어진 정도를 의미한다. 그래서 Q3 - Q1으로 IQR은 구해질 수 있는데, 

여기서 Q1은 데이터의 중앙값 아래에서의 중앙값을 의미하고, 

Q3은 데이터의 중앙값 위에서의 중앙값을 의미한다. 

쉽게 말해서 아래의 그림과 같이 중앙값은 데이터를 절반으로 했을 때의 값이라면, 

사분 범위는 1/4로 데이터를 쪼개어 중간에 두 구간을 나타내는 것이다. 




또한, 분기를 계산함에 있어서 아래와 같은 문제점이 있다. 


여기서는 두 가지를 제시하는데, 만약에 총 데이터의 개수가 홀수 개여서 정확하게 중앙값이 데이터에 속하는 경우이다. 이렇게 되면 그 데이터를 사분기를 정할 때 포함시킬 것인가 아닌가의 문제가 발생한다. 그래서 한 가지 방법으로는 중앙값을 그냥 무시하고 사분기를 설정하는 것과 나머지 한 가지 방법은 그냥 중앙값을 양쪽 사분기에 모두 포함하는 방법이 있다. 이것은 사용자의 입장에서 선택하는 문제인 것이고, 여기서는 중앙값을 제거하는 방법을 택했다고 한다. 

​

자, 그러면 이러한 정의를 바탕으로 정확하게 계산하는 예제를 풀어보자.


위의 예제에서는 stem plot을 이용해서 기대수명에 대한 자료를 국가별로 정리하였다. 여기서 stem plot이란, 아래와 같이, stem과 leaf로 나누어서 데이터를 나타내는 것인데, 왼쪽에 stem은 십의 자리라고 볼 수 있고, leaf은 일의 자리라고 해도 되겠다. 그러니 아래에서 보였듯이, 6 | 3 이라는 것은 63살이 되는 것이다. 

자 그러면 본래 우리의 예제(바로 위의 예제가 아닌, 주황색 음영된 데이터)에 대해서도 알 수 있을 것이다. 먼저 중앙값을 구해보자. 일의 자리 숫자 즉, leaf에서 숫자가 몇 개가 있는지가 총 데이터 개수가 될 것이다. 그래서 다 세어보면 총 23이 나오며 중앙값은 아래와 같이 12번째 값임이 구해지고, 이미 오름차순으로 정리된 데이터이므로, 12번째를 위에서부터 세어서 나타내면 73년이 된다. 


자, 그러면 IQR을 구해보도록 하자. IQR을 구하기 위해서는 우선적으로 Q1과 Q3 그리고 중앙값 M 값이 구해져야 한다. M은 앞에서 구했고, 이제 Q1과 Q3을 구해보자. 


앞서 이야기 했듯이, 여기서는 중앙값이 정확하게 데이터와 일치한다. 따라서, 해당 중앙값을 제외한 후에 Q1과 Q3을 구하는 방법을 택해서 구해보겠다. 중앙값을 제거하고 아래쪽에는 11개, 위쪽에는 11개의 데이터가 존재한다. Q1은 아래쪽 11개 데이터의 중앙값이고, Q3은 위쪽 11개의 중앙값이다.

그러면 (11+1)/2 인 6번째 값(66 years)이 Q1이 되고, (Q1 = 66), Q3은 23 + (11+1)/2 = 29번째 또는 중앙값에서 출발해서 6번째 값인 75 years가 된다. (Q3 = 75)

따라서, IQR = Q3 - Q1 = 75 - 66 = 9가 된다. 

R은 그냥 데이터의 범위(range)로 가장 큰 수와 가장 작은 수의 차이이므로 77 - 52 = 55가 된다. 

IQR가 9라고 했는데, 그 이야기는 중앙쪽에 위치한 데이터의 50%가 9만큼의 변화를 보인다는 것이고 이것이 표준편차와 비슷한 역할을 하여 데이터의 변화 정도를 측정하는 하나의 수치가 될 수 있다는 것을 알 수 있다. 즉 IQR 값이 클수록 표준 편차 값이 커지는 것 처럼 데이터가 많이 흩어져있을 가능성이 높은 것이다. 

그러면 IQR과 Median에 대해서 정확하게 이해했으리라 생각한다.
