```python
import json

from pathlib import Path

DB_FILE = Path("../json/sample.json")
```

```python
if not DB_FILE.exists():
    DB_FILE.write_text(json.dumps([]))
```

```python
messages = []

messages.append({"role" : "user", "content" : "안녕하세요"})
```

```python
json_text = json.dumps(messages, ensure_ascii=False, indent=2)
```

```python
DB_FILE.write_text(
    json_text,
    encoding="utf-8"
)
```

```python
# JSON 파일 내용을 문자열로 읽기
text = DB_FILE.read_text(encoding="utf-8")

print(text)
```

```python
json_text2 = json.dumps(messages, ensure_ascii=True, indent=2)

DB_FILE.write_text(
    json_text2,
    encoding="utf-8"
)
```

```python
text = DB_FILE.read_text(encoding="utf-8")

print(text)
```

```python
import json

data = json.loads(text)
print(type(data))
print(data)
```
