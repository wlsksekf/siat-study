```python
import os

from dotenv import load_dotenv

load_dotenv()

OPENAI_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_KEY is None:
    raise EnvironmentError(
        "openai api key 없음"
    )

print(f"OPENAI_API_KEY가 로드됨. 길이: {len(OPENAI_KEY)}")
```
